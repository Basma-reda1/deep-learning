```python
import numpy as np
np.random.seed(42)

tokens = ["What", "are", "the", "symptoms", "of", "diabetes", "?"]
seq_len = len(tokens)   
d_model  = 8          
d_k      = 4           

print("=" * 60)
print("INPUT SENTENCE: 'What are the symptoms of diabetes?'")
print(f"Tokens : {tokens}")
print(f"seq_len={seq_len}, d_model={d_model}, d_k={d_k}")
print("=" * 60)

X = np.random.randn(seq_len, d_model)  

W_Q = np.random.randn(d_model, d_k)    
W_K = np.random.randn(d_model, d_k)    
W_V = np.random.randn(d_model, d_k)    

Q = X @ W_Q     
K = X @ W_K     
V = X @ W_V     

print("\n── Query Matrix Q (7×4) ──")
print(np.round(Q, 3))

print("\n── Key Matrix K (7×4) ──")
print(np.round(K, 3))

scores_raw = Q @ K.T                  
scores_scaled = scores_raw / np.sqrt(d_k)   

print("\n── Raw Attention Scores  QKᵀ  (7×7) ──")
print(np.round(scores_raw, 3))

print("\n── Scaled Scores  QKᵀ/√d_k  (7×7) ──")
print(np.round(scores_scaled, 3))

def softmax(x, axis=-1):
    e_x = np.exp(x - np.max(x, axis=axis, keepdims=True))
    return e_x / e_x.sum(axis=axis, keepdims=True)

attention_weights = softmax(scores_scaled, axis=-1)   

print("\n── Attention Weights after Softmax (7×7) ──")
print(np.round(attention_weights, 4))

output = attention_weights @ V    

print("\n── Self-Attention Output  Z  (7×4) ──")
print(np.round(output, 4))

print("\n── Strongest Attention per Token ──")
print(f"{'Token':<12} {'Attends Most To':<15} {'Weight':>8}")
print("-" * 38)
for i, token in enumerate(tokens):
    j = np.argmax(attention_weights[i])
    print(f"{token:<12} {tokens[j]:<15} {attention_weights[i][j]:.4f}")
```
